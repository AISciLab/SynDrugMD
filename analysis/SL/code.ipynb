```python
import pandas as pd
df1 = pd.read_csv("ALL_Data_withGene1_withID_S.csv")
df2 = pd.read_csv("Human_SL.csv")
```

```python
df2
```

```python
df2_pairs = list(zip(df2['n1.identifier'], df2['n2.identifier']))
```

```python
df1
```

```python
df1_expanded = df1.copy()
df1_expanded['Anchor Target'] = df1_expanded['Anchor Target'].apply(lambda x: x.split(','))
df1_expanded['library Target'] = df1_expanded['library Target'].apply(lambda x: x.split(','))
df1_expanded
```

```python
def check_pairs(row, pairs):
    a_components = row['Anchor Target']
    b_components = row['library Target']
    for a_comp in a_components:
        for b_comp in b_components:
            if (a_comp, b_comp) in pairs or (b_comp, a_comp) in pairs:
                return 'y'
    return 'n'

# 添加匹配结果列
df1['match'] = df1_expanded.apply(check_pairs, axis=1, args=(df2_pairs,))
df1
```

```python
n_count = df1['match'].value_counts().get('n', 0)

print(f"不满足条件的行数: {n_count}")
```
